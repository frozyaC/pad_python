### Задание 1
- Создание объект Series из словаря, где ключи - имена городов, а значения - населения этих городов. Выведите этот Series.
- Создайте DataFrame из двух списков одинаковой длины, преобразовав его к DataFrame и
добавив собственные названия столбцов. Посмотрите на индексы и названия столбцов.
- Используя созданный DataFrame , выведите его методами df.info(), df.shape и
df.describe()

````python
import pandas as pd

# Создание Series из словаря
city_population = {'Москва': 12506468, 'Санкт-Петербург': 5351935, 'Новосибирск': 1620162, 'Екатеринбург': 1493749}
series_cities = pd.Series(city_population)
print("Series из словаря городов и их населения:")
print(series_cities)

# Создание DataFrame из двух списков
cities = ['Москва', 'Санкт-Петербург', 'Новосибирск', 'Екатеринбург']
populations = [12506468, 5351935, 1620162, 1493749]
df = pd.DataFrame(list(zip(cities, populations)), columns=['Город', 'Население'])
print("\nDataFrame: \n", df)

# Вывод индексов и названий столбцов
print("\nИндексы DataFrame:", df.index)
print("Названия столбцов DataFrame:", df.columns)

# Использование df.info(), df.shape, df.describe()
print("\nИнформация о DataFrame:")
df.info()
print("\nФорма DataFrame:", df.shape)
print("\nСтатистика по DataFrame:")
print(df.describe())
````


- Отфильтруйте df из примера по условию: Age > 30 и Score >= 90.
- Создайте DataFrame с покупками (столбцы: User , Item , Quantity ), затем посчитайте общее
количество покупок по каждому User с помощью groupby().
- Отсортируйте один из ваших DataFrame по нескольким столбцам (например, сначала по
Department , потом по Salary ).
- Используя groupby() , найдите средний возраст студентов в каждом классе (придумайте данные с
классами).

````python
# 1. Исходные DataFrame
df = pd.DataFrame({
    'Name': ['Alice','Bob','Charlie','David','Eva'],
    'Age': [25, 30, 35, 40, 22],
    'Score': [88, 92, 85, 90, 95]
})

df2 = pd.DataFrame({
    'Dept': ['HR','HR','IT','IT','HR'],
    'Salary': [100, 150, 200, 250, 180]
})

# Фильтрация по условию: Age > 30 и Score >= 90
filtered = df[(df['Age'] > 30) & (df['Score'] >= 90)]
print("Фильтрация df по условию Age > 30 и Score >= 90:")
print(filtered)
````

````python
# 2. DataFrame с покупками, и группировка по User
purchases = pd.DataFrame({
    'User': ['Alice','Bob','Alice','David','Bob','Charlie'],
    'Item': ['Book','Pen','Pencil','Book','Notebook','Pen'],
    'Quantity': [1,2,3,2,1,4]
})
total_purchases = purchases.groupby('User')['Quantity'].sum()
print("\nОбщее количество покупок по каждому User:")
print(total_purchases)
````

````python
# 3. Сортировка по нескольким столбцам (Dept, Salary)
sorted_df2 = df2.sort_values(by=['Dept', 'Salary'], ascending=[True, False])
print("\nСортировка df2 сначала по Dept, потом по Salary:")
print(sorted_df2)
````

````python
# 4. Пример данных с классами для поиска среднего возраста
students = pd.DataFrame({
    'Name': ['Ivan','Anna','Petr','Olga','Sergey','Maria'],
    'Age': [19, 21, 18, 22, 21, 20],
    'Class': ['A', 'B', 'A', 'B', 'A', 'B']
})
mean_age_by_class = students.groupby('Class')['Age'].mean()
print("\nСредний возраст по классам:")
print(mean_age_by_class)
````

### Задание 4

- В вашем DataFrame найдите количество пропущенных значений в каждом столбце (используйте
df.isna().sum()).
- Удалите все строки, где присутствуют пропуски во всех столбцах одновременно.
- Создайте столбец с датами в строковом формате (например, `df['date_str'] =
['2023-01-01', ...]` ), затем преобразуйте его в datetime и извлеките из него год и месяц.
- Приведите числовой столбец с плавающей точкой к целочисленному типу, а строковый столбец с
числами – к числовому типу с помощью pd.to_numeric()

````python
# Пример DataFrame с пропусками
df = pd.DataFrame({
    'A': [1, None, 3, None],
    'B': [None, None, 7, None],
    'C': [None, None, None, None],
    'D': [4.3, 2.7, None, None]
})

# 1. Найти количество пропущенных значений в каждом столбце
missing = df.isna().sum()
print("Количество пропущенных значений в каждом столбце:")
print(missing)
````

````python
# 2. Удалить все строки, где пропуски во всех столбцах одновременно
df_no_all_na = df.dropna(how='all')
print("\nDataFrame без строк, состоящих только из пропусков:")
print(df_no_all_na)
````

````python
# 3. Создаём столбец с датами в строковом формате, преобразуем в datetime и извлекаем год и месяц
df_no_all_na['date_str'] = ['2023-01-01', '2023-02-15', '2023-03-25', '2023-04-10'][:len(df_no_all_na)]
df_no_all_na['date'] = pd.to_datetime(df_no_all_na['date_str'])
df_no_all_na['year'] = df_no_all_na['date'].dt.year
df_no_all_na['month'] = df_no_all_na['date'].dt.month
print("\nDataFrame с датой, годом и месяцем:")
print(df_no_all_na[['date_str','date','year','month']])
````

````python
# 4. Преобразование столбца с float-числами к int, а строкового столбца с числами — к числовому типу
df_conv = pd.DataFrame({
    'float_col': [1.9, 2.7, 3.2],
    'str_num_col': ['10', '12.4', '15']
})
df_conv['float_col_int'] = df_conv['float_col'].astype(int)
df_conv['str_num_col_num'] = pd.to_numeric(df_conv['str_num_col'])
print("\nDataFrame с преобразованными типами:")
print(df_conv)
````
